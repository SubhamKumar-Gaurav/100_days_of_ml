# Titanic_without_using_pipeline

```python
import numpy as np
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer  # filling missing value 
from sklearn.preprocessing import OneHotEncoder  # for categorical - nominal 
from sklearn.preprocessing import MinMaxScaler   # Normalization 
from sklearn.tree import DecisionTreeClassifier  # ML Algo 
```

```python
df = pd.read_csv('titanic_train29.csv')
```

```python
df.shape
```

```python
uni = df['Pclass'].unique()
```

```python
unique_enties = df['Embarked'].unique()
```

```python
unique_enties
```

```python
df.drop(columns=['PassengerId',"Name",'Ticket','Cabin'], inplace=True )

# Dropping some columns 
```

```python
df
```

```python
df.shape
```

```python
# Step 1 -> train/test/split
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['Survived']),
                                                 df['Survived'],
                                                 test_size=0.2,
                                                random_state=42)
```

```python
X_train.head()
```

```python
y_train.head()
```

```python
df.isnull().sum() 
# There are some missing values in Embarked and Age 
```

```python
# Applying imputation

si_age = SimpleImputer()  # Replacing with mean 
si_embarked = SimpleImputer(strategy='most_frequent')  # replacing with high frequency 

X_train_age = si_age.fit_transform(X_train[['Age']])
X_train_embarked = si_embarked.fit_transform(X_train[['Embarked']])

X_test_age = si_age.transform(X_test[['Age']])
X_test_embarked = si_embarked.transform(X_test[['Embarked']])
```

```python
X_train_age
```

```python
X_train_embarked
```

```python
# one hot encoding on Sex and Embarked columns 

ohe_sex = OneHotEncoder(sparse_output=False,handle_unknown='ignore')
ohe_embarked = OneHotEncoder(sparse_output=False,handle_unknown='ignore')
# The reason why we are doing 2 separate ohe - look the above code of imputation

X_train_sex = ohe_sex.fit_transform(X_train[['Sex']])
X_train_embarked = ohe_embarked.fit_transform(X_train_embarked)

X_test_sex = ohe_sex.transform(X_test[['Sex']])
X_test_embarked = ohe_embarked.transform(X_test_embarked)
```

```python
X_train_sex
# There are two columns - one for male and another for female 
```

```python
X_train_embarked
```

```python
X_train_embarked.shape
```

```python
X_train.head(2)
```

### Dropping th old columns (which are now replaced by transformer ) 

```python
X_train_rem = X_train.drop(columns=['Sex','Age','Embarked']) 
```

```python
X_test_rem = X_test.drop(columns=['Sex','Age','Embarked'])
```

```python
X_test_rem
```

```python
X_train_transformed = np.concatenate((X_train_rem,X_train_age,X_train_sex,X_train_embarked),axis=1)
X_test_transformed = np.concatenate((X_test_rem,X_test_age,X_test_sex,X_test_embarked),axis=1)
```

```python
X_train_transformed.shape
```

```python
X_test_transformed.shape
```

### Decision Tree Classifier 

```python
clf = DecisionTreeClassifier()
clf.fit(X_train_transformed,y_train)
```

```python
y_pred = clf.predict(X_test_transformed)
y_pred
```

### Accuracy score 

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
import pickle 
```

```python
pickle.dump(ohe_sex,open('models/ohe_sex.pkl','wb'))
pickle.dump(ohe_embarked,open('models/ohe_embarked.pkl','wb'))
pickle.dump(clf,open('models/clf.pkl','wb')) 

# Exporting the files 
# Encoding in pkl format 
```
