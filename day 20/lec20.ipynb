# 20) EDA using Univariate Analysis

```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('titanic_train.csv')
```

```python
df.head()
```

## 1. Categorical Data

### a. Countplot

```python
# just counting the frequency in each column 
```

```python
df['Survived'].value_counts().plot(kind='bar') 
```

```python
df['Survived'].value_counts()
```

```python
df['Pclass'].value_counts().plot(kind='bar') 
```

```python
df['Pclass'].value_counts()
```

```python
df['Sex'].value_counts().plot(kind='bar') 
df['Sex'].value_counts()
```

```python
df['Embarked'].value_counts().plot(kind='bar') 
df['Embarked'].value_counts()
```

### b. PieChart

```python
# this is used to get the frequency in terms of percentage 
```

```python
df['Survived'].value_counts().plot(kind='pie' ,autopct='%.2f')
```

```python
df['Pclass'].value_counts().plot(kind='pie' ,autopct='%.2f')
```

```python
df['Sex'].value_counts().plot(kind='pie' ,autopct='%.2f')
```

## 2. Numerical Data 

### a. Histogram

```python
import matplotlib.pyplot as plt 
plt.hist(df["Age"])
# Show the exact number in continuous data 
```

```python
plt.hist(df["Age"],bins=5)
# with bins parameter, we can set the width 
```

```python
plt.hist(df["Age"],bins=20)
```

### b. Distplot

```python
sns.distplot(df['Age'])
# gives probability density function 
```

```python
sns.histplot(df['Age'])
```

### c. Boxplot

```python
sns.boxplot(df['Fare'])
# used when there is any noise or any outlier in the data 
```

```python
sns.boxplot(df['Fare'],orient='h') 
```

```python
sns.boxplot(df['Age'],orient='h') 
```
