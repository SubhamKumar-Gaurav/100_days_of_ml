# 26. Encoding Categorical Data

```python
import numpy as np 
import pandas as pd
```

```python
df=pd.read_csv("customer.csv")

# purchased column - we suggest a product after the user bought first product
```

```python
df.sample(5)
```

```python
df=df.iloc[:,2:]

# gender column is nominal and we need to do One Hot Encoding 
# review, education - Ordinal 
# purchased - label encoding 
```

```python
df.head() 
```

```python
df.shape
```

### Before we proceed further we need to split our data into train and test 

```python
from sklearn.model_selection import train_test_split 
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,0:2],df.iloc[:,-1],test_size=0.2)
```

```python
X_train
```

### Importing Ordinal Encoder class 

```python
from sklearn.preprocessing import OrdinalEncoder 
```

```python
# Values are given from min to max order (always) 
oe = OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']]) 
```

```python
oe.fit(X_train)

# fit - learn/determines what to assign 
# transform - actually assigns the value 
```

```python
X_train = oe.transform(X_train)
X_test = oe.transform(X_test)
```

```python
X_train

# We can see that the categorical data are converted to numerical (with ordering) 
```

```python
oe.categories_
```

### Importing Label Encoder class 

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
le.fit(y_train)
```

```python
le.classes_
```

```python
y_train = le.transform(y_train) 
y_test = le.transform(y_test)
```
