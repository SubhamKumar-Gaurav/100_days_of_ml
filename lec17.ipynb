## Making a request call to API 

```python
requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page=1')
```

## Saving the response that we got from API 

```python
response=requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page=1')
```

```python
# We got 4 key-value pairs , out of that 4 , we want 'results'
pd.DataFrame(response.json()['results']) 
```

```python
pd.DataFrame(response.json()['results'])[['id','title','overview','release_date','popularity','vote_average','vote_count']]
```

```python
df = pd.DataFrame(response.json()['results'])[['id','title','overview','release_date','popularity','vote_average','vote_count']]
```

```python
df
```

## Appending the entire data using pd.concat() fn.

```python
data = pd.DataFrame() 
```

```python
data 
```

```python
for i in range(1,429) :
    response=requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page={}'.format(i))
    df = pd.DataFrame(response.json()['results'])[['id','title','overview','release_date','popularity','vote_average','vote_count']]
    data=pd.concat([data,df],ignore_index=True) 
```

```python
data
```

# Saving the dataset 

```python
data.to_csv('movies_tmdb.csv')
```
